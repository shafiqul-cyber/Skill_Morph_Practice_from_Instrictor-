```python
# Upload the Gym data in google drive
gym = pd.read_csv('/content/drive/MyDrive/datasets/Assignment_4/HW/gym_membership.csv')
# Display last 5 rows
print(gym.tail())
# Check data types
print(gym.dtypes)
```

```python
# Task 1: Basic Matplotlib (3 plots)
# 1. Bar chart of gym membership types
# Use: gym['abonoment_type'].value_counts()
# Count gender
membership_count = gym['abonoment_type'].value_counts()
# Simple bar chart
plt.bar(['Standard', 'Premium'], membership_count.values, color=['tomato', 'blue'])
plt.title('Membership Count')
plt.grid(True, which="both", linestyle="-", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/barchargendervalue.png",dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 2. Histogram of time in gym
# Use: plt.hist(gym['avg_time_in_gym'])
# Average time distribution in gym.
plt.hist(gym['avg_time_in_gym'],bins=10, color='green')
plt.title('Average time Distribution in gym')
plt.grid(True, which="both", linestyle="-", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/math_scorehistogram.png",dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 3. Scatter plot: Age vs visits per week
# Use: plt.scatter(gym['Age'], gym['visit_per_week'])

from pickle import FALSE
# Math vs Reading

plt.scatter(gym['Age'], gym['visit_per_week'])
plt.grid(True)
plt.xlabel('Age')
plt.ylabel('Visit per week')
plt.title('age vs visit per week')
plt.grid(True, which="both", linestyle="-", linewidth=0.7, alpha=0.7)
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/age_vs_visit.png",dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 1. Box plot: Gender vs Age
# Use: sns.boxplot(data=gym, x='gender', y='Age')
sns.boxplot(data=gym, x='gender', y='Age')
plt.title('Age by Gender')
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/seaborngendervsage.png",dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 2. Count plot: Days per week
# Use: sns.countplot(data=gym, x='days_per_week')
sns.countplot(data=gym, x='days_per_week')
plt.title('Gym vs days per week')
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/gym_days_per_week.png",dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 3. Bar plot: Average visits by membership type
# Use: sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week')
plt.title('visit_per_week vs abonoment_type')
plt.savefig("/content/drive/MyDrive/datasets/Assignment_4/HW/visit_abnoment_type.png",dpi=300, bbox_inches='tight')
plt.show()
```
