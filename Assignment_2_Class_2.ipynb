## ***2.8 Create two sets of colors and find their union and intersection***

```python
# colors sets; Red Blue Green Yellow Orange Purple Black White Gray Brown

colors1={"Red","Blue","Green","Yellow","Orange","Purple"}
colors2={"Black","White","Gray","Brown","Blue","Green","Yellow"}
print(type(colors1))
print(type(colors2))
```

```python
# Union and intersection of these two sets of colors
union=colors1 | colors2 #union
inter=colors1 & colors2 #intersection
print("Union of sets of colors1 and colors2: ",union)
print("Intersection of sets of colors1 and colors2: ",inter)
```

## ***2.12 Write a program that prints a simple number pyramid pattern using nested loops.***

```python
#pattern 2
rows=5;
for i in range(1,rows+1):
  print(end=' '*(rows-i))
  for j in range(i):
    print(j+1, end=' ')
  print(' ')
```

## ***2.14 Create a function with default parameters that calculates the area of a rectangle.***

```python
# a function with default parameter

def calculate_area(length=int(input("Enter the default value of length:")),width=int(input("Enter the default value of width:"))):
  area=length*width
  return area
rectangle_area1=calculate_area() # calling the function without value, default value is taken
print("Calculate area without passing value:",rectangle_area1,"(m-m)")
rectangle_area2=calculate_area(12) # calling the function with only length value, width default value is taken
print("Calculate area calling with length only:",rectangle_area2,"(m-m)")
rectangle_area3=calculate_area(5,6) # calling the function with length and width value.
print("Calculate area calling with legth and width:",rectangle_area3,"(m-m)")
```
