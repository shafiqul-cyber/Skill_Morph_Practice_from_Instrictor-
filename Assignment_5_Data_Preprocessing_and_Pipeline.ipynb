```python
# Drop columns
df = df.drop(['mainroad', 'guestroom','basement','hotwaterheating','airconditioning','prefarea','furnishingstatus'], axis=1)
```

```python
print(df.columns.tolist())
```

```python
# Split into Train and Test
# TODO: Split - 80% train, 20% test
# Separate input (X) and output (y)
X = df.drop('price', axis=1)  # Everything except price
y = df['price']                # Only price

print("X has all  information except house price")
print("y has price (1 column)")

# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    random_state=20   # Same result every time
)

print(f"\nWe will train with: {len(X_train)} houses")
print(f"We will test with: {len(X_test)} houses")
```

```python
print(X_train.columns.tolist())
```

```python
#Apply StandardScaler
# Area is huge (1650-13300), Bedrooms is small (1-6)!
#Your Code here
# Create scaler
scaler = StandardScaler()

# Learn from training data and scale it
X_train_scaled = scaler.fit_transform(X_train)

# Scale test data (just transform, don't fit)
X_test_scaled = scaler.transform(X_test)

print("\nAfter scaling:")
print("All features now have similar range!")
print(f"Example - First house area before scaling: {X_train.values[0][0]}")
print(f"Example - First house area after scaling: {X_train_scaled[0][0]:.2f}")
```

```python
#train and test model
#your code here
from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(max_iter=10000) #LogisticRegression(max_iter=1000) → creates a logistic regression model that can iterate up to 1000 times during training to find the best solution.

# Train the model
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")
```

```python
# Check accuracy on training data
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")
```

```python
# Check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```

```python
from sklearn.model_selection import cross_val_score

# Do 5-fold cross-validation
scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=10            # 5 mini-tests
)

print("5 mini-test scores:")
for i in range(5):
    print(f"  Test {i+1}: {scores[i]:.1%}")

print(f"\nAverage: {scores.mean():.1%}")
print(f"This means our model is {scores.mean():.1%} accurate!")
```

```python
from sklearn.model_selection import cross_validate

cv_results = cross_validate(
    model,
    X_train_scaled,
    y_train,
    cv=10,
    return_train_score=True
)

print("Train Accuracies:", cv_results['train_score'])
print("Test Accuracies:", cv_results['test_score'])

print(f"\nAverage Train Accuracy: {cv_results['train_score'].mean():.1%}")
print(f"Average Test Accuracy:  {cv_results['test_score'].mean():.1%}")
```

```python
import joblib

# Save the model
joblib.dump(model, 'my_model.pkl')
print("✅ Model saved as 'my_model.pkl'")

# Save the scaler
joblib.dump(scaler, 'my_scaler.pkl')
print("✅ Scaler saved as 'my_scaler.pkl'")

# Save scaled data as CSV
pd.DataFrame(X_train_scaled).to_csv('train_data_scaled.csv', index=False)
print("✅ Scaled data saved as 'train_data_scaled.csv'")
```
